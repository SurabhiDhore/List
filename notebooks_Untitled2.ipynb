```python
list1 = []
list2 = []

print("Enter 5 values for list1:")
for i in range(5):
    element = int(input("Enter value:"))
    list1.append(element)

print("Enter 5 values for list2:")
for i in range(5):
    element = int(input("Enter value:"))
    list2.append(element)

print("List 1:", list1)
print("List 2:", list2)

data = list1 + list2

print("Merged list:", data)
```
